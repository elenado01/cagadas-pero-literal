```python
from pathlib import Path
import re

import pandas as pd


CHAT_PATH = Path("data/raw/Chat de WhatsApp con 💩.txt")
OUTPUT_PATH = Path("data/staging/cagadas.csv")

# WhatsApp export format: d/m/yy, h:mm - Full Name: message
message_pattern = re.compile(
    r"^(?P<dia>\d{1,2}/\d{1,2}/\d{2}),\s+"
    r"(?P<hora>\d{1,2}:\d{2})\s+-\s+"
    r"(?P<nombre>[^:]+):\s+(?P<mensaje>.*)$"
)

name_map = {
    "Elena Domingo": "Elena",
    "Lucía Erasmus": "Lucía",
    "Sol Erasmus": "Sol",
}

records = []
with CHAT_PATH.open(encoding="utf-8") as chat_file:
    for line in chat_file:
        match = message_pattern.match(line.rstrip("\n"))
        if match and match.group("mensaje") == "💩" and match.group("nombre") in name_map:
            records.append(
                {
                    "cagadora": name_map[match.group("nombre")],
                    "fecha": match.group("dia"),
                    "hora": match.group("hora"),
                }
            )

df = pd.DataFrame(records)
timestamps = pd.to_datetime(
    df["fecha"] + " " + df["hora"],
    format="%d/%m/%y %H:%M",
)
df["timestamp cagada"] = timestamps
df["fecha cagada"] = timestamps.dt.date
df["hora cagada"] = timestamps.dt.time
df["año"] = timestamps.dt.year
df["mes"] = timestamps.dt.month
df["día"] = timestamps.dt.day
df = df[
    [
        "cagadora",
        "timestamp cagada",
        "fecha cagada",
        "hora cagada",
        "año",
        "mes",
        "día",
    ]
]

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8-sig")

df
```
